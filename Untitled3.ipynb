```python

# ============================================================
# DATA MINING | ПРАКТИКАЛЫҚ ЖҰМЫС №9
# 3-НҰСҚА: HEART DISEASE CLASSIFICATION
# Орындаған: Айтказинов Арсен
# Тобы: АЖ-41
#
# Тақырып: Ensemble Learning — Stacking әдісі
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate
)
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import (
    GradientBoostingClassifier,
    StackingClassifier
)
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    RocCurveDisplay
)

RANDOM_STATE = 42

# ============================================================
# 1. ДЕРЕКТЕРДІ ЖҮКТЕУ
# ============================================================

print("=" * 65)
print("1. ДЕРЕКТЕРДІ ЖҮКТЕУ")
print("=" * 65)

URL = (
    "https://archive.ics.uci.edu/ml/"
    "machine-learning-databases/heart-disease/"
    "processed.cleveland.data"
)

COLUMNS = [
    "age", "sex", "cp", "trestbps",
    "chol", "fbs", "restecg", "thalach",
    "exang", "oldpeak", "slope",
    "ca", "thal", "num"
]

LOCAL = Path.cwd() / "processed.cleveland.data"

try:
    source = LOCAL if LOCAL.exists() else URL

    df = pd.read_csv(
        source,
        names=COLUMNS,
        na_values="?"
    )

    print("Деректер сәтті жүктелді!")

except Exception as error:
    raise RuntimeError(
        "Деректер жүктелмеді. Интернетті тексеріңіз "
        "немесе processed.cleveland.data файлын "
        "ноутбук орналасқан бумаға салыңыз."
    ) from error

print("\nДеректер өлшемі:", df.shape)
print("\nАлғашқы 5 жол:")
print(df.head())

# ============================================================
# 2. ДЕРЕКТЕРДІ ТАЛДАУ
# ============================================================

print("\n" + "=" * 65)
print("2. ДЕРЕКТЕРДІ ТАЛДАУ")
print("=" * 65)

for column in COLUMNS:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print("\nБағандар:")
print(df.columns.tolist())

print("\nДеректер түрлері:")
print(df.dtypes)

print("\nЖетіспейтін мәндер:")
print(df.isnull().sum())

print("\nЖетіспейтін мәндердің жалпы саны:")
print(df.isnull().sum().sum())

print("\nСтатистикалық сипаттама:")
print(df.describe().round(2))

# ============================================================
# 3. МАҚСАТТЫ АЙНЫМАЛЫНЫ ДАЙЫНДАУ
# ============================================================

print("\n" + "=" * 65)
print("3. МАҚСАТТЫ АЙНЫМАЛЫ")
print("=" * 65)

# UCI деректерінде:
# num = 0 -> жүрек ауруы анықталмаған
# num = 1,2,3,4 -> ауру белгісі бар

df = df.dropna(subset=["num"]).copy()

df["target"] = (df["num"] > 0).astype(int)

print("\nКласс саны:")
print(df["target"].value_counts().sort_index())

print("\nКласс үлесі (%):")
print(
    (
        df["target"].value_counts(
            normalize=True
        ).sort_index() * 100
    ).round(2)
)

plt.figure(figsize=(7, 4))

counts = df["target"].value_counts().sort_index()

plt.bar(
    ["Ауру жоқ (0)", "Ауру бар (1)"],
    counts.values
)

plt.title("Мақсатты кластардың таралуы")
plt.ylabel("Объект саны")
plt.tight_layout()
plt.show()

# ============================================================
# 4. БЕЛГІЛЕРДІ ДАЙЫНДАУ
# ============================================================

print("\n" + "=" * 65)
print("4. БЕЛГІЛЕРДІ ДАЙЫНДАУ")
print("=" * 65)

X = df.drop(
    columns=["num", "target"]
)

y = df["target"]

numeric_features = [
    "age",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak"
]

categorical_features = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]

numeric_transformer = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "scaler",
        StandardScaler()
    )
])

categorical_transformer = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])

preprocessor = ColumnTransformer([
    (
        "numeric",
        numeric_transformer,
        numeric_features
    ),
    (
        "categorical",
        categorical_transformer,
        categorical_features
    )
])

print("Сандық белгілер:", numeric_features)
print("Категориялық белгілер:", categorical_features)

# ============================================================
# 5. TRAIN / TEST БӨЛУ
# ============================================================

print("\n" + "=" * 65)
print("5. TRAIN / TEST БӨЛУ")
print("=" * 65)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Train өлшемі:", X_train.shape)
print("Test өлшемі:", X_test.shape)

print("\nTrain кластар үлесі:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTest кластар үлесі:")
print(y_test.value_counts(normalize=True).round(3))

# ============================================================
# 6. БАЗАЛЫҚ МОДЕЛЬДЕР
# ============================================================

print("\n" + "=" * 65)
print("6. БАЗАЛЫҚ МОДЕЛЬДЕР")
print("=" * 65)

lr = Pipeline([
    ("preprocessor", preprocessor),
    (
        "classifier",
        LogisticRegression(
            max_iter=2000,
            random_state=RANDOM_STATE
        )
    )
])

gb = Pipeline([
    ("preprocessor", preprocessor),
    (
        "classifier",
        GradientBoostingClassifier(
            n_estimators=100,
            learning_rate=0.1,
            max_depth=3,
            random_state=RANDOM_STATE
        )
    )
])

print("\nLogistic Regression оқытылуда...")
lr.fit(X_train, y_train)

print("Gradient Boosting оқытылуда...")
gb.fit(X_train, y_train)

print("Екі базалық модель оқытылды.")

# ============================================================
# 7. STACKING МОДЕЛІ
# ============================================================

print("\n" + "=" * 65)
print("7. STACKING МОДЕЛІ")
print("=" * 65)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)

stack = StackingClassifier(
    estimators=[
        (
            "lr",
            Pipeline([
                ("preprocessor", preprocessor),
                (
                    "classifier",
                    LogisticRegression(
                        max_iter=2000,
                        random_state=RANDOM_STATE
                    )
                )
            ])
        ),
        (
            "gb",
            Pipeline([
                ("preprocessor", preprocessor),
                (
                    "classifier",
                    GradientBoostingClassifier(
                        n_estimators=100,
                        learning_rate=0.1,
                        max_depth=3,
                        random_state=RANDOM_STATE
                    )
                )
            ])
        )
    ],
    final_estimator=LogisticRegression(
        max_iter=2000,
        random_state=RANDOM_STATE
    ),
    cv=cv,
    stack_method="predict_proba"
)

print("Stacking оқытылуда...")
stack.fit(X_train, y_train)

print("Stacking моделі сәтті оқытылды.")

# ============================================================
# 8. МОДЕЛЬ САПАСЫН БАҒАЛАУ
# ============================================================

print("\n" + "=" * 65)
print("8. МОДЕЛЬ САПАСЫН БАҒАЛАУ")
print("=" * 65)

models = {
    "Logistic Regression": lr,
    "Gradient Boosting": gb,
    "Stacking": stack
}

results = []
predictions = {}

for name, model in models.items():

    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    test_prob = model.predict_proba(
        X_test
    )[:, 1]

    predictions[name] = test_pred

    results.append({
        "Model": name,
        "Train Accuracy": accuracy_score(
            y_train, train_pred
        ),
        "Test Accuracy": accuracy_score(
            y_test, test_pred
        ),
        "Precision": precision_score(
            y_test, test_pred,
            zero_division=0
        ),
        "Recall": recall_score(
            y_test, test_pred,
            zero_division=0
        ),
        "Macro-F1": f1_score(
            y_test, test_pred,
            average="macro",
            zero_division=0
        ),
        "ROC-AUC": roc_auc_score(
            y_test, test_prob
        )
    })

results_df = pd.DataFrame(results)

print("\nМодельдерді салыстыру:")
print(
    results_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# ============================================================
# 9. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 65)
print("9. CLASSIFICATION REPORT")
print("=" * 65)

for name in models:

    print(f"\n{name}")

    print(
        classification_report(
            y_test,
            predictions[name],
            target_names=[
                "Ауру жоқ",
                "Ауру бар"
            ],
            zero_division=0
        )
    )

# ============================================================
# 10. CONFUSION MATRIX
# ============================================================

print("\n" + "=" * 65)
print("10. CONFUSION MATRIX")
print("=" * 65)

for name in models:

    cm = confusion_matrix(
        y_test,
        predictions[name],
        labels=[0, 1]
    )

    tn, fp, fn, tp = cm.ravel()

    print(f"\n{name}")
    print("TN:", tn)
    print("FP:", fp)
    print("FN:", fn)
    print("TP:", tp)

    print(
        "FN — ауруы бар адамды "
        "сау деп қате анықтау."
    )

    print(
        "FP — сау адамды "
        "ауру деп қате анықтау."
    )

fig, axes = plt.subplots(
    1, 3,
    figsize=(15, 4)
)

for ax, (name, model) in zip(
    axes, models.items()
):

    ConfusionMatrixDisplay.from_estimator(
        model,
        X_test,
        y_test,
        display_labels=["Сау", "Ауру"],
        cmap="Blues",
        ax=ax,
        colorbar=False
    )

    ax.set_title(name)

plt.tight_layout()
plt.show()

# ============================================================
# 11. ROC CURVE
# ============================================================

print("\n" + "=" * 65)
print("11. ROC CURVE")
print("=" * 65)

fig, ax = plt.subplots(figsize=(8, 6))

for name, model in models.items():

    RocCurveDisplay.from_estimator(
        model,
        X_test,
        y_test,
        name=name,
        ax=ax
    )

ax.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    color="gray"
)

ax.set_title("ROC қисықтарын салыстыру")
plt.tight_layout()
plt.show()

# ============================================================
# 12. МЕТРИКАЛАРДЫ САЛЫСТЫРУ
# ============================================================

print("\n" + "=" * 65)
print("12. МЕТРИКАЛАРДЫ САЛЫСТЫРУ")
print("=" * 65)

plot_data = results_df.set_index("Model")[
    [
        "Test Accuracy",
        "Recall",
        "Macro-F1",
        "ROC-AUC"
    ]
]

ax = plot_data.plot(
    kind="bar",
    figsize=(11, 5),
    ylim=(0, 1.08)
)

ax.set_title(
    "Logistic Regression, Gradient Boosting "
    "және Stacking салыстыру"
)

ax.set_ylabel("Метрика мәні")
ax.set_xlabel("Модель")

plt.xticks(rotation=15, ha="right")
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()

# ============================================================
# 13. OVERFITTING ТАЛДАУЫ
# ============================================================

print("\n" + "=" * 65)
print("13. OVERFITTING ТАЛДАУЫ")
print("=" * 65)

results_df["Accuracy Gap"] = (
    results_df["Train Accuracy"]
    - results_df["Test Accuracy"]
)

for _, row in results_df.iterrows():

    print("\nМодель:", row["Model"])

    print(
        "Train Accuracy:",
        round(row["Train Accuracy"], 4)
    )

    print(
        "Test Accuracy:",
        round(row["Test Accuracy"], 4)
    )

    print(
        "Айырмашылық:",
        round(row["Accuracy Gap"], 4)
    )

    if row["Accuracy Gap"] > 0.10:
        print(
            "Train/Test айырмашылығы үлкен. "
            "Overfitting қаупі байқалады."
        )
    else:
        print(
            "Train/Test айырмашылығы "
            "салыстырмалы түрде шағын."
        )

# ============================================================
# 14. CROSS-VALIDATION
# ============================================================

print("\n" + "=" * 65)
print("14. CROSS-VALIDATION")
print("=" * 65)

cv_results = []

for name, model in models.items():

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=StratifiedKFold(
            n_splits=5,
            shuffle=True,
            random_state=RANDOM_STATE
        ),
        scoring={
            "accuracy": "accuracy",
            "f1_macro": "f1_macro",
            "recall": "recall",
            "roc_auc": "roc_auc"
        },
        n_jobs=1
    )

    cv_results.append({
        "Model": name,
        "CV Accuracy": scores[
            "test_accuracy"
        ].mean(),
        "CV Macro-F1": scores[
            "test_f1_macro"
        ].mean(),
        "CV Recall": scores[
            "test_recall"
        ].mean(),
        "CV ROC-AUC": scores[
            "test_roc_auc"
        ].mean(),
        "CV Std": scores[
            "test_accuracy"
        ].std()
    })

cv_df = pd.DataFrame(cv_results)

print(
    cv_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# ============================================================
# 15. НӘТИЖЕЛЕРДІ САҚТАУ
# ============================================================

print("\n" + "=" * 65)
print("15. НӘТИЖЕЛЕРДІ САҚТАУ")
print("=" * 65)

results_df.to_csv(
    "variant3_model_results.csv",
    index=False,
    encoding="utf-8-sig"
)

cv_df.to_csv(
    "variant3_cross_validation.csv",
    index=False,
    encoding="utf-8-sig"
)

print("CSV файлдары сақталды.")

# ============================================================
# 16. АВТОМАТТЫ ҚОРЫТЫНДЫ
# ============================================================

print("\n" + "=" * 65)
print("16. ҚОРЫТЫНДЫ")
print("=" * 65)

best_f1 = results_df.loc[
    results_df["Macro-F1"].idxmax()
]

best_recall = results_df.loc[
    results_df["Recall"].idxmax()
]

best_auc = results_df.loc[
    results_df["ROC-AUC"].idxmax()
]

print(
    "\nMacro-F1 бойынша үздік модель:",
    best_f1["Model"]
)

print(
    "Macro-F1 мәні:",
    round(best_f1["Macro-F1"], 4)
)

print(
    "\nRecall бойынша үздік модель:",
    best_recall["Model"]
)

print(
    "Recall мәні:",
    round(best_recall["Recall"], 4)
)

print(
    "\nROC-AUC бойынша үздік модель:",
    best_auc["Model"]
)

print(
    "ROC-AUC мәні:",
    round(best_auc["ROC-AUC"], 4)
)



print("\nЖАЛПЫ ҚОРЫТЫНДЫ")
print("=" * 65)

print(
    "Бұл практикалық жұмыста жүрек ауруын анықтау "
    "үшін үш модель салыстырылды:"
)

print("1. Logistic Regression")
print("2. Gradient Boosting")
print("3. Stacking")

print(
    "\nMacro-F1 бойынша үздік модель:",
    best_f1["Model"]
)

print(
    "Macro-F1:",
    round(best_f1["Macro-F1"], 4)
)

print(
    "\nRecall бойынша үздік модель:",
    best_recall["Model"]
)

print(
    "Recall:",
    round(best_recall["Recall"], 4)
)

print(
    "\nROC-AUC бойынша үздік модель:",
    best_auc["Model"]
)

print(
    "ROC-AUC:",
    round(best_auc["ROC-AUC"], 4)
)

print(
    "\nМедициналық есепте Recall маңызды, "
    "себебі ауруы бар адамды сау деп қате "
    "анықтау қауіпті болуы мүмкін."
)

print(
    "\nStacking бірнеше модельдің болжамдарын "
    "біріктіреді, бірақ оның нәтижесі жеке "
    "модельдерден әрқашан жоғары бола бермейді."
)

print(
    "\nҚорытынды: модельдердің тиімділігі "
    "Accuracy, Recall, Macro-F1 және ROC-AUC "
    "метрикалары арқылы бағаланды. "
    "Бұл нәтижелер тек оқу мақсатында "
    "қолданылады және медициналық диагноз "
    "қоюға арналмаған."
)

print("\nПрактикалық жұмыс №9 аяқталды!")
```
